```python
import pandas as pd

# 1. Load the dataset
df = pd.read_csv('netflix_titles.csv')

# 2. Handle missing values
# Fill missing text data with 'Unknown'
df['director'] = df['director'].fillna('Unknown')
df['cast'] = df['cast'].fillna('Unknown')
df['country'] = df['country'].fillna('Unknown')
# Drop the few remaining rows where critical data like date_added or rating is missing
df = df.dropna(subset=['date_added', 'rating', 'duration'])

# 3. Standardize text and formats
# Ensure headers are clean (lowercase, no spaces)
df.columns = df.columns.str.lower().str.replace(' ', '_')
# Remove leading/trailing spaces from the date column
df['date_added'] = df['date_added'].str.strip()

# 4. Remove any duplicate rows
df = df.drop_duplicates()

# 5. Export the cleaned dataset
df.to_csv('cleaned_netflix_data.csv', index=False)
print("Data cleaned and saved successfully!")
```
